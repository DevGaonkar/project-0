```python
# Extracting text from images and analyzing the sentiment

# Install necessary libraries
!pip install easyocr --quiet
import easyocr
import cv2
from matplotlib import pyplot as plt
import numpy as np

# Reading the image and Optical Character Recognition
path = 'image.png'

reader = easyocr.Reader(['en'])  # Initialize the easyocr reader
result = reader.readtext(path)    # Extract text from the image
result
```

```python
# Extracting the coordinates of the detected text
top_left = tuple(result[0][0][0])  # top-left corner
bottom_right = tuple(result[0][0][2])  # bottom-right corner
text = result[0][1]  # The extracted text
font = cv2.FONT_HERSHEY_DUPLEX  # Font style

# Drawing a rectangle and displaying the extracted text on the image
img = cv2.imread(path)
img = cv2.rectangle(img, top_left, bottom_right, (0, 255, 0), 2)  # Green rectangle
font = cv2.FONT_HERSHEY_SIMPLEX
img = cv2.putText(img, text, top_left, font, 1, (255, 255, 255), 2, cv2.LINE_AA)  # White text

# Show the image with rectangle and text
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()

# Extract the specific detected text
single_text = result[0][1]
print(single_text)  # Output: 'Hello Dev!'
```

```python
# Reading the image with multiple lines of text
path2 = 'mimage.png'

# Initialize the easyocr reader
reader = easyocr.Reader(['en'])
result = reader.readtext(path2)  # Extract text from the image

# Reading the image using OpenCV for drawing bounding boxes
img = cv2.imread(path2)
spacer = 100

# Concatenate the detected words into one single text
full_text = " ".join([detection[1] for detection in result])

# Drawing bounding boxes and displaying text on the image (with thinner border)
for detection in result:
    top_left = tuple(int(val) for val in detection[0][0])
    bottom_right = tuple(int(val) for val in detection[0][2])
    text = detection[1]

    # Change thickness from 3 to 2 for a thinner border
    img = cv2.rectangle(img, top_left, bottom_right, (0, 255, 0), 1)  # Thinner border (thickness = 2)
    img = cv2.putText(img, text, (top_left[0], top_left[1] - spacer),
                      cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 255, 0), 2, cv2.LINE_AA)

spacer = 15
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.show()

# Print the concatenated text
print("Full Extracted Text:", full_text)
```

```python
# Install necessary libraries
!pip install textblob --quiet
!python -m textblob.download_corpora

from textblob import TextBlob
import nltk

# Example text
single_text = "Hello Dev!"  # Ensure this is set

# Create a TextBlob object
blob1 = TextBlob(single_text)

# Part-of-speech tagging (this will give tags for each word)
print("POS Tags:", blob1.tags)

# Extract noun phrases (useful for identifying key nouns)
print("Noun Phrases:", blob1.noun_phrases)

# Perform sentiment analysis (this gives polarity and subjectivity)
print("Sentiment:", blob1.sentiment)
```
