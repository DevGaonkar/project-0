```python
from copy import deepcopy

def checkElementHelper(x, S):
    """Return membership value of element x in fuzzy set S."""
    for e in S:
        if x == e[1]:
            return e[0]
    return 0

# Define fuzzy sets
ClassA = [[0.7,"Shivpratik"], [0.2,"Bhushan"], [0.4,"Sahil"], [0.6,"Atharva"], [0.3,"Rohit"]]
ClassB = [[0.8,"Shivpratik"], [0.4,"Bhushan"], [0.7,"Sahil"], [0.1,"Atharva"], [0.5,"rohit"]]

def union(A, B):
    """Fuzzy Union: μA∪B(x) = max(μA(x), μB(x))"""
    X, Y = deepcopy(A), deepcopy(B)
    Z = []
    for i in X:
        mb = checkElementHelper(i[1], Y)
        Z.append([max(mb, i[0]), i[1]])
        if mb != 0:
            Y.remove([mb, i[1]])
    Z += Y
    return Z
print("Union:", union(ClassA, ClassB))

def intersection(A, B):
    """Fuzzy Intersection: μA∩B(x) = min(μA(x), μB(x))"""
    X, Y = deepcopy(A), deepcopy(B)
    Z = []
    for i in X:
        mb = checkElementHelper(i[1], Y)
        Z.append([min(mb, i[0]), i[1]])
        if mb != 0:
            Y.remove([mb, i[1]])
    return Z
print("Intersection:", intersection(ClassA, ClassB))

def complement(A):
    """Fuzzy Complement: μA'(x) = 1 - μA(x)"""
    X = deepcopy(A)
    Z = []
    for i in X:
        Z.append([1 - i[0], i[1]])
    return Z
print("Complement ClassA:", complement(ClassA))


# def cartesian_product(A, B):
#     """Fuzzy Cartesian Product: μA×B(x,y) = min(μA(x), μB(y))"""
#     X, Y = deepcopy(A), deepcopy(B)
#     Z = []
#     for i in X:
#         for j in Y:
#             Z.append([min(i[0], j[0]), (i[1], j[1])])
#     return Z
# print("Cartesian Product:", cartesian_product(ClassA, ClassB))
# --- EXTRA OPERATIONS ---

def fuzzy_sum(A, B):
    """Algebraic Sum: μA+B(x) = μA(x) + μB(x) - μA(x)μB(x)"""
    X, Y = deepcopy(A), deepcopy(B)
    Z = []
    for i in X:
        mb = checkElementHelper(i[1], Y)
        Z.append([i[0] + mb - (i[0] * mb), i[1]])
        if mb != 0:
            Y.remove([mb, i[1]])
    Z += Y
    return Z
print("Fuzzy Sum:", fuzzy_sum(ClassA, ClassB))

def fuzzy_product(A, B):
    """Algebraic Product: μA*B(x) = μA(x) * μB(x)"""
    X, Y = deepcopy(A), deepcopy(B)
    Z = []
    for i in X:
        mb = checkElementHelper(i[1], Y)
        Z.append([i[0] * mb, i[1]])
        if mb != 0:
            Y.remove([mb, i[1]])
    return Z
print("Fuzzy Product:", fuzzy_product(ClassA, ClassB))

def bounded_sum(A, B):
    """Bounded Sum: μA⊕B(x) = min(1, μA(x) + μB(x))"""
    X, Y = deepcopy(A), deepcopy(B)
    Z = []
    for i in X:
        mb = checkElementHelper(i[1], Y)
        Z.append([min(1, i[0] + mb), i[1]])
        if mb != 0:
            Y.remove([mb, i[1]])
    Z += Y
    return Z
print("Bounded Sum:", bounded_sum(ClassA, ClassB))

def bounded_difference(A, B):
    """Bounded Difference: μA⊖B(x) = max(0, μA(x) - μB(x))"""
    X, Y = deepcopy(A), deepcopy(B)
    Z = []
    for i in X:
        mb = checkElementHelper(i[1], Y)
        Z.append([max(0, i[0] - mb), i[1]])
        if mb != 0:
            Y.remove([mb, i[1]])
    return Z
print("Bounded Difference:", bounded_difference(ClassA, ClassB))





```
